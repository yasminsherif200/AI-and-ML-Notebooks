# 1. Data Preparation

```python
from tensorflow import keras

# Load dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Flatten images (28x28 → 784)
x_train_flat = x_train.reshape(-1, 28 * 28).astype('float32') / 255.0
x_test_flat = x_test.reshape(-1, 28 * 28).astype('float32') / 255.0

print("Training shape:", x_train_flat.shape)
print("Test shape:", x_test_flat.shape)
```

# 2. Experiment with different Activation Functions

# 2.1 Relu Activation Function

```python
from tensorflow import keras
from tensorflow.keras import layers

model1 = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(784,)),   # hidden layer 1
    layers.Dense(32, activation='relu'),                        # hidden layer 2
    layers.Dense(10, activation='softmax')                      # output (10 classes)
])



model1.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# model1.summary()

# Train the model
history1 = model1.fit(
    x_train_flat, y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)

# Evaluate the model
loss1, acc1 = model1.evaluate(x_test_flat, y_test)

print("Model 1 Test Accuracy:", acc1)
```

Test Accuracy: 0.9714
- Observations:
  - The model performs very well on MNIST (~97% accuracy)
  - ReLU activation helps fast convergence

# 2.2 Tanh Activation Function

```python
model2 = keras.Sequential([
    layers.Dense(64, activation='tanh', input_shape=(784,)),
    layers.Dense(32, activation='tanh'),
    layers.Dense(10, activation='softmax')
])

model2.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train
history2 = model2.fit(
    x_train_flat, y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)

# Evaluate
loss2, acc2 = model2.evaluate(x_test_flat, y_test)

print("Model 2 Test Accuracy:", acc2)
```

Test Accuracy: 0.9682
- Observations:

    - Accuracy is slightly lower than the when we used ReLU activation function (0.9714 → 0.9682)

    - Training is still stable, but learning is a bit less efficient than ReLU

# 3. Experiment with Learning Rate

we will keep the same model but we will change the optimizer

```python
from tensorflow.keras.optimizers import Adam

model3 = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(784,)),
    layers.Dense(32, activation='relu'),
    layers.Dense(10, activation='softmax')
])

optimizer = Adam(learning_rate=0.0001)
model3.compile(
    optimizer=optimizer,
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history3 = model3.fit(
    x_train_flat, y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)

loss3, acc3 = model3.evaluate(x_test_flat, y_test)

print("Model 3 Test Accuracy:", acc3)
```

Test Accuracy: 0.9438
- Observations:
    - This model uses a smaller learning rate (0.0001)
    - Training is more stable but significantly slower to converge
    - Accuracy is lower than previous models (0.9714 → 0.9438)
    - The model learns more cautiously but underperforms in limited epochs

# 4. Final Comparison of All Models

| Model | Activation | Learning Rate | Test Accuracy |
|------|------------|---------------|---------------|
| Model 1 | ReLU | 0.001 (default) | 0.9714 |
| Model 2 | Tanh | 0.001 (default) | 0.9682 |
| Model 3 | ReLU | 0.0001 | 0.9438 |

# Conclusion:
- ReLU performs better than Tanh in this dataset
- Default learning rate (0.001) is the most effective for this problem
- Reducing learning rate too much slows learning and reduces performance within limited epochs
